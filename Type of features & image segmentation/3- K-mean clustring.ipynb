```python
# Load an image
image = cv2.imread('pattern.png')
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(rgb_image)
```

```python
# Reshape image into a 2D array and 3 color values
image_copy = rgb_image.copy()
pixel_values = image_copy.reshape((-1, 3))

# Convert to float type
pixel_values = np.float32(pixel_values)
```

```python
# Define stopping criteria
ceriteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 10, 1.0)
```

```python
# K mean clustering
k = 3
retval, labels, centers = cv2.kmeans(pixel_values, k, None, ceriteria, 10, cv2.KMEANS_RANDOM_CENTERS)
```

```python
# Convert data into 8-bit values
centers = np.uint8(centers)
segmented_data = centers[labels.flatten()]

# Reshape data into the orginal image dimentations
segmented_image = segmented_data.reshape((image.shape))
label_reshape = labels.reshape(image.shape[0], image.shape[1])

# Plot the result
plt.imshow(segmented_image)
```
