```python
# Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load an image
image = cv2.imread('fruits.avif')
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Plot the result
plt.imshow(rgb_image)
```

```python
# Convert to grayscale
gray_image = cv2.cvtColor(image.copy(), cv2.COLOR_BGR2GRAY)

# Convert to float32
gray_image = np.float32(gray_image)
```

```python
# Detect corners
gray_corner = cv2.cornerHarris(gray_image, 2, 29, 0.06)
gray_corner = cv2.dilate(gray_corner, None)

# Plot corners
plt.imshow(gray_corner, cmap = 'gray')
```
