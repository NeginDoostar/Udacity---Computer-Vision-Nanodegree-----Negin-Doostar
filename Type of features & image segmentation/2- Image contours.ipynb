```python
# Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load an image
image = cv2.imread('3flowers.jpeg')
rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Plot the result
plt.imshow(rgb_image)
```

```python
# Convert to grayscale
gray_image = cv2.cvtColor(image.copy(), cv2.COLOR_BGR2GRAY)

# Make a threshold to mask image
retval, binary_image = cv2.threshold(gray_image, 225, 255, cv2.THRESH_BINARY)

# Plot the mask
plt.imshow(binary_image, cmap = 'gray')
```

```python
# Find contours
contours, hierachy = cv2.findContours(binary_image, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

# Draw contours
all_contours = cv2.drawContours(rgb_image.copy(), contours, -1, (255, 0, 0), 2)

# Plot the result
plt.imshow(all_contours)
```
