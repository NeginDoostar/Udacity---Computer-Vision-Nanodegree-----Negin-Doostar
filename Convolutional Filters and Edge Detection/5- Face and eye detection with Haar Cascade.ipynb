```python
# Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load a face image
face = cv2.imread('face.jpg')
face_rgb = cv2.cvtColor(face, cv2.COLOR_BGR2RGB)

# Plot the image
plt.imshow(face_rgb)
```

```python
# Detecting face

# Load classifer
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

face_copy = face_rgb.copy()
face_rect = face_cascade.detectMultiScale(face_copy, scaleFactor = 1.2, minNeighbors = 5)

for (x, y, w, h) in face_rect:
    cv2.rectangle(face_copy, (x, y), (x + w, y + h), (255, 255, 255), 10)

# Plot the result
plt.imshow(face_copy)
```

```python
# Detecting eyes

# Load classifer
eye_cascade = cv2.CascadeClassifier('haarcascade_eye.xml')

eye_copy = face_rgb.copy()
eye_rect = eye_cascade.detectMultiScale(eye_copy, scaleFactor = 1.2, minNeighbors = 10)

for (x, y, w, h) in eye_rect:
    cv2.rectangle(eye_copy, (x, y), (x + w, y + h), (255, 255, 255), 10)

# Plot the result
plt.imshow(eye_copy)
```
